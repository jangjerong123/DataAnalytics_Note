## Open API

API(Application Programming Interface) : 웹 환경 HTTP기반- URL에 요청(request)을 보내면 서버가 그에 맞는 data를 돌려주는(response) 방식
- 가장 널리 사용되는 방식 : REST(Representataional State Transfer) : 약자 알아두기 vs Web 스크래핑(무단복제-HTML화면 결과 출력)
- SOAP : 단어만 기억
Web API
- GET(read),POST(create),PUT(update),DELETE(delete) : 꼭 알아두기
- data 존재-> CRUD(Create,Read,Update,Delete) 

String vs Path 방식 in REST API: 매우 중요
- String : URL끝에 ?를 붙이고 &로 Key=value형태로 연결하여 값 전달
- Path : GET 사용/ URL경로 자체에 값 포

```python
%%writefile src/mylib.py
def sayHello():
    print('hello')
def getKey(KeyPath):
    d=dict()
    f=open(KeyPath,'r') #file이 없으면 오류
    for line in f.readlines(): #readline은 한 줄, readlines는 전체
        row=line.split('=')
        row0=row[0]
        d[row0]=row[1].strip() #strip():뒤에 공백 제거 #d{'ds','숫자'}
    f.close()
    return d
        
```
