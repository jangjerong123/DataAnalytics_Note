### Map Reduce

```python
# map-reduce 강조!
# word cloud에 많이 사용, 반복문이 없음(행렬 사용)
# lambda랑 많이 사용 - 함수 선언이 한 줄 짜리기 때문
# **Map함수 func 자리의 함수를 전에 미리 선언하지 않고 lambda를 사용해서 Map()함수 내부에서 선언 및 사용 가능**
# map(fucn,iterable) : iterable의 각 원소에 func를 적용한 결과들의 모음(이터레이터) 반환
# 마지막엔 list로 반환해야 함 - map()는 map object로 반환되기 때문
def c2f(c):
    return (float(9)/5)*c+32
celsius=[39.2,36.5,37.3,37.8]
f=map(c2f,celsius)
print(list(f))
```

### Map+Lambda

```python
# lamnda함수 인자 2개 가능한 거 강조
x=lambda x,y:x+y
print(x(1,2))

# 아래 코드 매우매우 강조 (lambda와 map을 이용해서 코드 쓸 줄 알아여 함)
# 반복문이 없어도 map()함수 안에 0부터 n까지 쭉 진행
f=map(lambda c:(float(9)/5)*c+32,celsius)
print(list(f))

#아래 두 코드 차이 알아두자
list(map(lambda x:x.split(),"Hello World"))

sentence=["Hello World", "Good Morning"]
list(map(lambda x:x.split(),sentence))
```

### Reduce

```python
#Reduce(func,iterable):data개수만큼 결과값을 반환하는 map()과 달리 reduce()는 하나의 결과값만 반환 
from functools import reduce
reduce(lambda x,y:x+y,range(1,101))
```
